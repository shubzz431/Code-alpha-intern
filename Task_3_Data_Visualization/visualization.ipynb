## Task 3: Data Visualization (Simple & Clean)

In this notebook, we load the books dataset and create 4 simple visualizations:
1. **Bar Chart**: Top 10 Book Categories
2. **Histogram**: Book Price Distribution
3. **Pie Chart**: Star Rating Breakdown
4. **Box Plot**: Price vs. Star Rating

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
```

### 1. Load Data

```python
df = pd.read_csv('books_data.csv')
print(f"Total Books Loaded: {len(df)}")
df[['title', 'category', 'price_gbp', 'rating', 'units_available']].head()
```

### 2. Chart 1: Top 10 Book Categories (Bar Chart)

```python
top_cats = df['category'].value_counts().head(10)

plt.figure(figsize=(9, 4.5))
bars = plt.barh(top_cats.index[::-1], top_cats.values[::-1], color='#3b82f6', edgecolor='black')
plt.title('Top 10 Book Categories', fontsize=13, fontweight='bold')
plt.xlabel('Number of Books')
for bar in bars:
    w = bar.get_width()
    plt.text(w + 1, bar.get_y() + bar.get_height()/2, f"{int(w)}", va='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 3. Chart 2: Book Price Distribution (Histogram)

```python
plt.figure(figsize=(9, 4.5))
plt.hist(df['price_gbp'], bins=20, color='#10b981', edgecolor='black', alpha=0.8)
avg = df['price_gbp'].mean()
plt.axvline(avg, color='red', linestyle='--', label=f"Average Price (£{avg:.2f})")
plt.title('Distribution of Book Prices (£ GBP)', fontsize=13, fontweight='bold')
plt.xlabel('Price (£)')
plt.ylabel('Book Count')
plt.legend()
plt.tight_layout()
plt.show()
```

### 4. Chart 3: Star Rating Share (Pie Chart)

```python
ratings = df['rating'].value_counts().sort_index()
colors = ['#f87171', '#fb923c', '#facc15', '#60a5fa', '#4ade80']
labels = [f"{r} Star ({cnt})" for r, cnt in ratings.items()]

plt.figure(figsize=(6, 6))
plt.pie(ratings.values, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, wedgeprops=dict(edgecolor='white', linewidth=2))
plt.title('Customer Star Rating Share', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5. Chart 4: Price vs Star Rating (Box Plot)

```python
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df, x='rating', y='price_gbp', palette='Set2')
plt.title('Price Distribution Across Star Ratings', fontsize=13, fontweight='bold')
plt.xlabel('Star Rating (1 to 5)')
plt.ylabel('Price (£ GBP)')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import json

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
```

### 1. Load Processed Dataset

```python
df = pd.read_csv('../data/processed/books_cleaned.csv')
print(f"Total Books: {len(df)}")
print(f"Columns ({len(df.columns)}): {list(df.columns)}")
df.head()
```

### 2. Dataset Schema & Missing Value Check

```python
df.info()
print("\nMissing values per column:")
print(df.isnull().sum())
```

### 3. Summary Statistics

```python
df[['price_gbp', 'star_rating', 'stock_quantity']].describe()
```

### 4. Pricing Distribution

```python
plt.figure(figsize=(9, 5))
plt.hist(df['price_gbp'], bins=25, color='#3498db', edgecolor='#2980b9', alpha=0.75)
plt.axvline(df['price_gbp'].mean(), color='#e74c3c', linestyle='--', label=f"Mean: £{df['price_gbp'].mean():.2f}")
plt.axvline(df['price_gbp'].median(), color='#2ecc71', linestyle='-.', label=f"Median: £{df['price_gbp'].median():.2f}")
plt.title('Distribution of Book Prices (£ GBP)', fontsize=14, fontweight='bold')
plt.xlabel('Price (£)')
plt.ylabel('Book Count')
plt.legend()
plt.tight_layout()
plt.show()
```

### 5. Star Rating Frequency

```python
ratings = df['star_rating'].value_counts().sort_index()
plt.figure(figsize=(8, 4.5))
bars = plt.bar([f"{r} Star" for r in ratings.index], ratings.values, color='#f39c12', edgecolor='#d68910', width=0.5)
for b in bars:
    h = b.get_height()
    plt.annotate(f"{h}", (b.get_x() + b.get_width()/2, h), textcoords="offset points", xytext=(0, 4), ha='center', fontweight='bold')
plt.title('Star Rating Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

### 6. Top Book Categories

```python
top_cats = df['category'].value_counts().head(10).sort_values(ascending=True)
plt.figure(figsize=(10, 5))
top_cats.plot(kind='barh', color='#1abc9c', edgecolor='#16a085')
plt.title('Top 10 Book Categories by Count', fontsize=14, fontweight='bold')
plt.xlabel('Number of Books')
plt.tight_layout()
plt.show()
```
